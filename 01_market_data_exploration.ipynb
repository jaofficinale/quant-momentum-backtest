```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
import quantstats as qs

plt.style.use("seaborn-v0_8-whitegrid")
```

```python
tickers =[
    "SPY",
    "QQQ",
    "IWM",
    "TLT",
    "GLD",
    "VNQ",
]
```

```python
prices = yf.download(
    tickers,
    start="2015-01-01",
    end="2026-08-01",
    auto_adjust=True,
    progress=False
    )["Close"]

prices = prices.dropna()

prices.head()
```

```python
from pathlib import Path
import os

print("JupyterLab is currently running from:")
print(os.getcwd())

data_path = Path("./data")
data_path.mkdir(parents=True, exist_ok=True)

file_path= data_path/"etf_prices_daily.csv"

prices.to_csv(file_path)
print("\nSuccesfully saved file to:")
print(file_path.resolve())

print("\nDataset shape:", prices.shape)
print("Data range:", prices.index.min().date(), "to", prices.index.max().date())
```
